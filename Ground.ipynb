```python
@dataclass
class Message(MessageBase):
    device_name: str = None
    mac_address: int = None
    is_encrypted: int = None
    content_type: int = None
    content: MessageBase = None

    __sizes__ = {
        "device_name": 64,
        "mac_address": 6,
        "is_encrypted": 1,
        "content_type": 1,
    }

t = Message()
t.device_name = "waseem"
t.__annotations__
```
